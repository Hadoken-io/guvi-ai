```python
d={'username':['aniket@gmail.com', '1234','ques','ans1','ques2','ans2']}
sq=['place?','pet?','school?','food?','movie?']
```

```python
def checkpassword(password):
    ac=dc=uc=lc=sc=0
    strength=0
    spch="!@#$%^&*()-_=+[{]}\|;:'\",<.>/?"
    for char in password:
        if char.isalpha():
            ac+=1
        if char.isupper():
            uc+=1
        elif char.islower():
            lc+=1
        elif char.isdigit():
            dc+=1
        elif char in spch:
            sc+=1
    if len(password)>=8 :
        strength+=1
    if ac and dc:
        strength+=1
    if uc and lc:
        strength+=1
    if sc:
        strength+=1
    if strength==4:
        return password
    else:
        print("Password is weak, please enter a strong password")
        password=input("Enter your password: ")
        return checkpassword(password)
```

```python
def sequirityquestion():
    from random import choice as c
    temp= list(range(0,5))
    I1 = c(temp)
    temp.remove(I1)
    I2 = c(temp)
    q1 = sq[I1]
    q2 = sq[I2]
    a='n'
    while a=='n':
        ans1 = input(f'{q1}: ')
        ans2 = input(f'{q2}: ')
        a=input('do u want to save answers? (y/n): ')
    return q1, ans1, q2, ans2
```

```python
def dbupdate(username, email, password, q1, ans1, q2, ans2):
    d[username] = [email, password, q1, ans1, q2, ans2]
    return True
```

```python
def signup():
    username = input("Enter a username: ").lower()
    if username not in d:
        email = input("Enter your email: ")
        #verifyemail()
        password = input("Enter a password: ")
        password=checkpassword(password)
        q1,ans1,q2,ans2=sequirityquestion()
        dbupdate(username,email,password,q1,ans1,q2,ans2)
        
        print("Signup successful!")
    else:
        print("user exists")
        signup()

```

**encryption and decryption using ciser cipher** 
*homework*
- 2 factor authentication

```python
def caesar_encrypt(text, shift=3):
    result = ""
    for char in text:
        if char.isalpha():
            if char.isupper():
                result += chr((ord(char) - ord('A') + shift) % 26 + ord('A'))
            else:
                result += chr((ord(char) - ord('a') + shift) % 26 + ord('a'))
        elif char.isdigit():
            result += str((int(char) + shift) % 10)
        else:
            result += char
    return result


def caesar_decrypt(text, shift=3):
    result = ""
    for char in text:
        if char.isalpha():
            if char.isupper():
                result += chr((ord(char) - ord('A') - shift + 26) % 26 + ord('A'))
            else:
                result += chr((ord(char) - ord('a') - shift + 26) % 26 + ord('a'))
        elif char.isdigit():
            result += str((int(char) - shift + 10) % 10)
        else:
            result += char
    return result
```

```python
msg = "Hello123"
enc = caesar_encrypt(msg, shift=3)
dec = caesar_decrypt(enc, shift=3)
print("Original:", msg)
print("Encrypted:", enc)
print("Decrypted:", dec)
```

**2 Factor Authentication (2FA)**


```python
import random


def generate_otp(length=6):
    otp = ""
    for _ in range(length):
        otp += str(random.randint(0, 9))
    return otp


def verify_otp(sent_otp, user_input=None):
    if user_input is None:
        user_input = input("Enter the OTP sent to your registered contact: ")
    if user_input == sent_otp:
        return True
    else:
        return False


def two_factor_auth():
    otp = generate_otp()
    print(f"[2FA] OTP sent: {otp} (For demo purposes)")
    if verify_otp(otp):
        print("2FA verification successful!")
        return True
    else:
        print("2FA verification failed!")
        return False
```

```python
# Demo 2FA
two_factor_auth()
```
